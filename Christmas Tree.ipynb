**A Christmas Trie**

This notebook shows the main features of a christmas trie.

```python
from system.utils import Vector, Triangle
from ipycanvas import Canvas, hold_canvas

x_width: int = 120
y_width: int = x_width
t: Triangle = Triangle(Vector(0, 0), Vector(x_width, 0), Vector(x_width / 2, x_width))


def recursive_draw(
    t: Triangle, canvas: Canvas, canvas_height: int, rec_depth: int, x_offset: int = 0
):
    """Draws recursive triangles."""
    if rec_depth == 0:
        return
    t.draw(canvas, canvas_height, x_offset, y_offset=0)
    subs = t.split_into_sub_descriptors()
    for t_sub in subs:
        recursive_draw(t_sub, canvas, canvas_height, rec_depth - 1, x_offset)


# draw sub-triangles for different depths:
max_rec_level: int = 8
canvas = Canvas(width=x_width * max_rec_level, height=y_width, sync_image_data=True)

for rec_depth in range(1, max_rec_level):
    with hold_canvas(canvas):
        recursive_draw(t, canvas, y_width, rec_depth, x_offset=rec_depth * x_width)
display(canvas)
```

```python
# check and visualize containment in triangle:
canvas = Canvas(width=x_width, height=y_width, sync_image_data=True)
with hold_canvas(canvas):
    display(canvas)
    recursive_draw(t, canvas, y_width, 1)
    for i in range(100):
        for x in range(0, 300, 7):
            for y in range(0, 300, 7):
                contained: bool = t.contains(Vector(x, y))
                if contained:
                    canvas.fill_style = "#FFFFFF"
                else:
                    canvas.fill_style = "#ff0066"

                canvas.fill_rect(x, y_width - y, 2, 2)
```

**Animated tree:**

```python
number_of_trees: int = 20
size: int = 300
canvas = Canvas(width=size * number_of_trees, height=size, sync_image_data=True)
import random as rnd
from time import sleep

triangle: Triangle = Triangle(Vector(0, 0), Vector(size, 0), Vector(size / 2, size))
christmas_trie: ChristmasTree[Vector, str] = ChristmasTree[Vector, str](
    children_per_inner_node=4,
    descriptor=triangle,
    number_of_inner_node_levels=3,
    # inner_node_factory=ChristmasTree.InnerNodeFactory[Vector, str](config="inner")
    # inner_node_factory=ChristmasTree.InnerNodeFactory[Vector, str](config="buffered")
    inner_node_factory=ChristmasTree.InnerNodeFactory[Vector, str](
        config="crystalball"
    ),
)

display(canvas)

for tree in range(number_of_trees):
    with hold_canvas(canvas):
        canvas.clear()
        for i in range(size):
            v: Vector
            while True:
                v = Vector(rnd.randint(0, size), rnd.randint(0, size))
                if triangle.contains(v):
                    break
            christmas_trie.put(v, str(i))
        christmas_trie.draw(canvas, canvas_height=size, x_offset=0)
        sleep(0.2)
```

**Each phase of the tree shown separately:**

```python
number_of_trees: int = 5
size: int = 200
space_in_between: int = 20
canvas = Canvas(
    width=size * number_of_trees + space_in_between * (number_of_trees - 1),
    height=size,
    sync_image_data=True,
)
import random as rnd
from time import sleep

triangle: Triangle = Triangle(Vector(0, 0), Vector(size, 0), Vector(size / 2, size))
christmas_trie: ChristmasTree[Vector, str] = ChristmasTree[Vector, str](
    children_per_inner_node=4,
    descriptor=triangle,
    number_of_inner_node_levels=3,
    # inner_node_factory=ChristmasTree.InnerNodeFactory[Vector, str](config="inner")
    # inner_node_factory=ChristmasTree.InnerNodeFactory[Vector, str](config="buffered")
    inner_node_factory=ChristmasTree.InnerNodeFactory[Vector, str](
        config="crystalball"
    ),
)

display(canvas)

for tree in range(number_of_trees):
    with hold_canvas(canvas):
        # canvas.clear()
        for i in range(1000):
            v: Vector
            while True:
                v = Vector(rnd.randint(0, size), rnd.randint(0, size))
                if triangle.contains(v):
                    break
            christmas_trie.put(v, str(i))
        christmas_trie.draw(
            canvas, canvas_height=size, x_offset=tree * size + space_in_between * tree
        )
        # sleep(0.2)
```
