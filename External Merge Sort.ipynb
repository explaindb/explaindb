```python
from faker import Faker

fake = Faker()
fake.seed_instance(42)

from dataclasses import dataclass

number_of_tuples = 1 * 10 * 1000


@dataclass
class Stuff:
    a: int
    b: int


# create some fake data mimicking tuples in a relation:
data = [
    Stuff(fake.pyint(max_value=number_of_tuples), fake.pyint(max_value=2000))
    for f in range(number_of_tuples)
]
data[:10]
```

**Algorithm 1: Sort in main memory**

```python
data_sorted = sorted(data, key=lambda x: x.a)
```

```python
data_sorted[:10]
```

**Algorithm 2: External Merge Sort**

```python
# Assume we have a "large" file opn disk/SSD:
# "large" meaning: file is too large to fit into main memory
import pickle

# write data into file to emulate store
with open("stuff.pkl", "wb") as f:
    pickle.dump([stuff for stuff in data], f)
```
